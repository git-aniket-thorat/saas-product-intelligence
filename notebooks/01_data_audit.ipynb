Task 2 — Investigate missing values Missing-value analysis

```python
missing_summary = []

for name, df in datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        missing_percentage = (
            missing_count / len(df) * 100
            if len(df) > 0
            else 0
        )

        missing_summary.append({
            "dataset": name,
            "column": column,
            "missing_count": missing_count,
            "missing_percentage": round(
                missing_percentage, 2
            )
        })

missing_df = pd.DataFrame(missing_summary)

# Show only columns that contain missing values
missing_df = missing_df[
    missing_df["missing_count"] > 0
].sort_values(
    by=["dataset", "missing_count"],
    ascending=[True, False]
)

display(missing_df.style.hide(axis="index"))
```

```python
duplicate_row_summary = []

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    duplicate_row_summary.append({
        "dataset": name,
        "total_rows": len(df),
        "duplicate_rows": duplicate_count
    })

duplicate_rows_df = pd.DataFrame(
    duplicate_row_summary
)

display(duplicate_rows_df)
```

Check primary-key uniqueness

```python
primary_keys = {
    "accounts": "account_id",
    "subscriptions": "subscription_id",
    "features": "usage_id",
    "support": "ticket_id",
    "churn": "churn_event_id"
}

primary_key_summary = []

for name, key in primary_keys.items():

    df = datasets[name]

    duplicate_count = df[key].duplicated().sum()

    missing_key_count = df[key].isna().sum()

    primary_key_summary.append({
        "dataset": name,
        "primary_key": key,
        "duplicate_key_count": duplicate_count,
        "missing_key_count": missing_key_count,
        "primary_key_valid": (
            duplicate_count == 0
            and missing_key_count == 0
        )
    })

primary_key_df = pd.DataFrame(
    primary_key_summary
)

display(primary_key_df)
```

Investigate the 21 repeated feature usage IDs

```python
features = datasets["features"]

# Identify all rows whose usage_id occurs more than once
duplicate_usage_rows = features[
    features.duplicated(
        subset=["usage_id"],
        keep=False
    )
].sort_values("usage_id")

print(
    "Number of rows associated with repeated usage IDs:",
    len(duplicate_usage_rows)
)

display(duplicate_usage_rows)
```

```python
duplicate_usage_counts = (
    duplicate_usage_rows
    .groupby("usage_id")
    .size()
    .reset_index(name="occurrence_count")
    .sort_values(
        "occurrence_count",
        ascending=False
    )
)

display(duplicate_usage_counts)
```

```python
duplicate_usage_comparison = (
    duplicate_usage_rows
    .groupby("usage_id")
    .agg(
        occurrence_count=("usage_id", "size"),
        unique_subscriptions=("subscription_id", "nunique"),
        unique_usage_dates=("usage_date", "nunique"),
        unique_feature_names=("feature_name", "nunique"),
        unique_usage_counts=("usage_count", "nunique")
    )
    .reset_index()
)

display(duplicate_usage_comparison)
```

Validate the relationships between tables

```python
orphaned=~datasets["subscriptions"]["account_id"].isin(datasets["accounts"]["account_id"])
print(orphaned.sum())
```

```python
def find_orphan_records(
    child_df,
    child_key,
    parent_df,
    parent_key
):
    orphan_mask=~child_df[child_key].isin(parent_df[parent_key])
    return child_df.loc[orphan_mask].copy()
```

```python
relationships = [
    {
        "relationship": "Subscriptions -> Accounts",
        "child": "subscriptions",
        "child_key": "account_id",
        "parent": "accounts",
        "parent_key": "account_id"
    },
    {
        "relationship": "Feature usage -> Subscriptions",
        "child": "features",
        "child_key": "subscription_id",
        "parent": "subscriptions",
        "parent_key": "subscription_id"
    },
    {
        "relationship": "Support tickets -> Accounts",
        "child": "support",
        "child_key": "account_id",
        "parent": "accounts",
        "parent_key": "account_id"
    },
    {
        "relationship": "Churn events -> Accounts",
        "child": "churn",
        "child_key": "account_id",
        "parent": "accounts",
        "parent_key": "account_id"
    }
]

relationship_results = []

orphan_records = {}

for relationship in relationships:

    child_df = datasets[relationship["child"]]
    parent_df = datasets[relationship["parent"]]

    orphan_df = find_orphan_records(
        child_df,
        relationship["child_key"],
        parent_df,
        relationship["parent_key"]
    )

    orphan_records[relationship["relationship"]] = orphan_df

    relationship_results.append({
        "relationship": relationship["relationship"],
        "child_records": len(child_df),
        "orphan_records": len(orphan_df),
        "relationship_valid": len(orphan_df) == 0
    })

relationship_df = pd.DataFrame(
    relationship_results
)

display(relationship_df)
```

Validate date columns

```python
date_columns = {
    "accounts": ["signup_date"],
    "subscriptions": ["start_date", "end_date"],
    "features": ["usage_date"],
    "support": ["submitted_at", "closed_at"],
    "churn": ["churn_date"]
}
```

```python
date_audit_datasets = {
    name: df.copy()
    for name, df in datasets.items()
}

date_parse_results = []

for name, columns in date_columns.items():

    df = date_audit_datasets[name]

    for column in columns:

        original_values = df[column]

        parsed_dates = pd.to_datetime(
            original_values,
            errors="coerce"
        )

        invalid_date_mask = (
            original_values.notna()
            & parsed_dates.isna()
        )

        date_parse_results.append({
            "dataset": name,
            "column": column,
            "non_missing_values": original_values.notna().sum(),
            "invalid_date_count": invalid_date_mask.sum(),
            "invalid_date_examples": (
                original_values[invalid_date_mask]
                .astype(str)
                .head(5)
                .tolist()
            )
        })

        df[f"{column}_parsed"] = parsed_dates

date_parse_df = pd.DataFrame(
    date_parse_results
)

display(date_parse_df)
```

Check whether subscription end dates precede start dates end date< start date

```python
subscriptions_audit = date_audit_datasets["subscriptions"]

invalid_subscription_periods = subscriptions_audit[
    subscriptions_audit["end_date_parsed"].notna()
    &
    (
        subscriptions_audit["end_date_parsed"]
        <
        subscriptions_audit["start_date_parsed"]
    )
]

print(
    "Subscriptions ending before they start:",
    len(invalid_subscription_periods)
)

display(invalid_subscription_periods.head(20))
```

Check support ticket timestamps

```python
support_audit = date_audit_datasets["support"]

invalid_support_periods = support_audit[
    support_audit["closed_at_parsed"].notna()
    &
    support_audit["submitted_at_parsed"].notna()
    &
    (
        support_audit["closed_at_parsed"]
        <
        support_audit["submitted_at_parsed"]
    )
]

print(
    "Tickets closed before submission:",
    len(invalid_support_periods)
)

display(invalid_support_periods.head(20))
```

Compare recorded resolution time with timestamps

```python
support_time_check = support_audit.copy()

support_time_check["calculated_resolution_hours"] = (
    (
        support_time_check["closed_at_parsed"]
        -
        support_time_check["submitted_at_parsed"]
    ).dt.total_seconds()
    / 3600
)

support_time_check["resolution_difference_hours"] = (
    support_time_check["resolution_time_hours"]
    -
    support_time_check["calculated_resolution_hours"]
)

support_time_check["resolution_mismatch"] = (
    support_time_check["resolution_difference_hours"]
    .abs() > 0.01
)

resolution_mismatches = support_time_check[
    support_time_check["resolution_mismatch"]
]

print(
    "Tickets with a resolution-time mismatch:",
    len(resolution_mismatches)
)

display(
    resolution_mismatches[
        [
            "ticket_id",
            "submitted_at",
            "closed_at",
            "resolution_time_hours",
            "calculated_resolution_hours",
            "resolution_difference_hours"
        ]
    ].head(20)
)
```

Check whether churn occurred before account signup

```python
accounts_audit = date_audit_datasets["accounts"]
churn_audit = date_audit_datasets["churn"]

churn_with_signup = churn_audit.merge(
    accounts_audit[
        ["account_id", "signup_date_parsed"]
    ],
    on="account_id",
    how="left",
    validate="many_to_one"
)

invalid_churn_dates = churn_with_signup[
    churn_with_signup["churn_date_parsed"].notna()
    &
    churn_with_signup["signup_date_parsed"].notna()
    &
    (
        churn_with_signup["churn_date_parsed"]
        <
        churn_with_signup["signup_date_parsed"]
    )
]

print(
    "Churn events before account signup:",
    len(invalid_churn_dates)
)

display(invalid_churn_dates.head(20))
```

Validate numerical values

```python
numeric_checks = []

def record_numeric_check(
    dataset_name,
    column,
    rule,
    invalid_df
):

    numeric_checks.append({
        "dataset": dataset_name,
        "column": column,
        "rule": rule,
        "invalid_record_count": len(invalid_df)
    })
```

```python
# Accounts: seat count should be positive
accounts = datasets["accounts"]

record_numeric_check(
    "accounts",
    "seats",
    "Seats must be greater than zero",
    accounts[accounts["seats"] <= 0]
)


# Subscriptions: seat count should be positive
subscriptions = datasets["subscriptions"]

record_numeric_check(
    "subscriptions",
    "seats",
    "Seats must be greater than zero",
    subscriptions[subscriptions["seats"] <= 0]
)


# Feature usage: usage count should not be negative
features = datasets["features"]

record_numeric_check(
    "features",
    "usage_count",
    "Usage count must not be negative",
    features[features["usage_count"] < 0]
)


# Feature usage: duration should not be negative
record_numeric_check(
    "features",
    "usage_duration_secs",
    "Usage duration must not be negative",
    features[features["usage_duration_secs"] < 0]
)


# Feature usage: error count should not be negative
record_numeric_check(
    "features",
    "error_count",
    "Error count must not be negative",
    features[features["error_count"] < 0]
)
```

Check financial values

```python
# Monthly recurring revenue should not be negative
record_numeric_check(
    "subscriptions",
    "mrr_amount",
    "MRR must not be negative",
    subscriptions[subscriptions["mrr_amount"] < 0]
)


# Annual recurring revenue should not be negative
record_numeric_check(
    "subscriptions",
    "arr_amount",
    "ARR must not be negative",
    subscriptions[subscriptions["arr_amount"] < 0]
)


# Refund amounts should not be negative
churn = datasets["churn"]

record_numeric_check(
    "churn",
    "refund_amount_usd",
    "Refund amount must not be negative",
    churn[churn["refund_amount_usd"] < 0]
)
```

Check support response time and satisfaction scores

```python
support = datasets["support"]

record_numeric_check(
    "support",
    "resolution_time_hours",
    "Resolution time must not be negative",
    support[support["resolution_time_hours"] < 0]
)

record_numeric_check(
    "support",
    "first_response_time_minutes",
    "First response time must not be negative",
    support[support["first_response_time_minutes"] < 0]
)

invalid_satisfaction = support[
    support["satisfaction_score"].notna()
    &
    (
        (support["satisfaction_score"] < 1)
        |
        (support["satisfaction_score"] > 5)
    )
]

record_numeric_check(
    "support",
    "satisfaction_score",
    "Non-missing satisfaction scores must be between 1 and 5",
    invalid_satisfaction
)
```

Review numerical validation results

```python
numeric_check_df = pd.DataFrame(numeric_checks)

display(
    numeric_check_df.sort_values(
        by=["invalid_record_count", "dataset"],
        ascending=[False, True]
    )
)
```

Validate subscription business rules

```python
subscriptions = datasets["subscriptions"]

subscription_end_date_review = (
    subscriptions
    .groupby(
        ["churn_flag", "is_trial"],
        dropna=False
    )
    .agg(
        total_subscriptions=("subscription_id", "size"),
        missing_end_dates=("end_date", lambda s: s.isna().sum())
    )
    .reset_index()
)

subscription_end_date_review["missing_end_date_percentage"] = (
    subscription_end_date_review["missing_end_dates"]
    /
    subscription_end_date_review["total_subscriptions"]
    * 100
).round(2)

display(subscription_end_date_review)
```

Create an audit results summary

```python
audit_summary = []

# Dataset sizes
for name, df in datasets.items():

    audit_summary.append({
        "category": "Dataset overview",
        "dataset": name,
        "check": "Dataset loaded",
        "issue_count": 0,
        "status": "PASS"
    })


# Missing values
for name, df in datasets.items():

    missing_count = int(df.isna().sum().sum())

    audit_summary.append({
        "category": "Completeness",
        "dataset": name,
        "check": "Missing values",
        "issue_count": missing_count,
        "status": (
            "REVIEW"
            if missing_count > 0
            else "PASS"
        )
    })


# Duplicate rows
for name, df in datasets.items():

    duplicate_count = int(df.duplicated().sum())

    audit_summary.append({
        "category": "Duplicates",
        "dataset": name,
        "check": "Exact duplicate rows",
        "issue_count": duplicate_count,
        "status": (
            "REVIEW"
            if duplicate_count > 0
            else "PASS"
        )
    })


# Primary-key uniqueness
for name, key in primary_keys.items():

    df = datasets[name]

    duplicate_count = int(
        df[key].duplicated().sum()
    )

    missing_key_count = int(
        df[key].isna().sum()
    )

    issue_count = duplicate_count + missing_key_count

    audit_summary.append({
        "category": "Identifiers",
        "dataset": name,
        "check": f"Primary key: {key}",
        "issue_count": issue_count,
        "status": (
            "REVIEW"
            if issue_count > 0
            else "PASS"
        )
    })


# Foreign-key relationships
for result in relationship_results:

    audit_summary.append({
        "category": "Relationships",
        "dataset": result["relationship"],
        "check": "Foreign-key integrity",
        "issue_count": result["orphan_records"],
        "status": (
            "REVIEW"
            if result["orphan_records"] > 0
            else "PASS"
        )
    })


# Date parsing
for result in date_parse_results:

    audit_summary.append({
        "category": "Dates",
        "dataset": result["dataset"],
        "check": f"Date parsing: {result['column']}",
        "issue_count": result["invalid_date_count"],
        "status": (
            "REVIEW"
            if result["invalid_date_count"] > 0
            else "PASS"
        )
    })


# Numerical checks
for result in numeric_checks:

    audit_summary.append({
        "category": "Numerical validation",
        "dataset": result["dataset"],
        "check": f"{result['column']}: {result['rule']}",
        "issue_count": result["invalid_record_count"],
        "status": (
            "REVIEW"
            if result["invalid_record_count"] > 0
            else "PASS"
        )
    })


audit_summary_df = pd.DataFrame(audit_summary)

display(
    audit_summary_df.sort_values(
        by=["category", "status", "dataset"]
    )
)
```
